```python
df["Price"] = df["Price_in_Lakhs"] * 100000
```

```python
df["Price_per_SqFt"] = df["Price"] / df["Size_in_SqFt"]
```

```python
for col in df.select_dtypes(include="object").columns:
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
for col in df.select_dtypes(include="str").columns:
    print(col, df[col].str.strip().equals(df[col]))
```

```python
df["Future_Price_5Y"] = df["Price"] * (1 + 0.08) ** 5
```

```python
df[["Price", "Future_Price_5Y"]].head()
```

```python
median_price_sqft = df["Price_per_SqFt"].median()
median_price_sqft
```

```python
df["Price_SqFt_Score"] = (
    df["Price_per_SqFt"] <= median_price_sqft
).astype(int)
```

```python
df["Price_SqFt_Score"].value_counts()
```

```python
df["BHK_Score"] = (df["BHK"] >= 3).astype(int)
```

```python
df["BHK_Score"].value_counts()
```

```python
df["Ready_Move_Score"] = (
    df["Availability_Status"] == "Ready_to_Move"
).astype(int)
```

```python
df["Ready_Move_Score"].value_counts()
```

```python
df["Investment_Score"] = (
    df["Price_SqFt_Score"]
    + df["BHK_Score"]
    + df["Ready_Move_Score"]
)
```

```python
df["Investment_Score"].value_counts().sort_index()
```

```python
df["Good_Investment"] = (
    df["Investment_Score"] >= 2
).astype(int)
```

```python
df["Good_Investment"].value_counts()
```

```python
df.drop(
    columns=[
        "Price_SqFt_Score",
        "BHK_Score",
        "Ready_Move_Score",
        "Investment_Score"
    ],
    inplace=True
)
```

```python
df["Amenities"].unique()
```

```python
amenities = set()

for value in df["Amenities"]:
    for amenity in value.split(","):
        amenities.add(amenity.strip())

print(amenities)
print("Total amenities:", len(amenities))
```

```python
amenities = ["Garden", "Gym", "Playground", "Clubhouse", "Pool"]

for amenity in amenities:
    df[f"Amenity_{amenity}"] = (
        df["Amenities"].str.contains(amenity, regex=False)
    ).astype(int)
```
